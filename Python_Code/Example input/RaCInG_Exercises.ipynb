## 1. Getting Started

As a first step, we need to read the `TOY` data. All data needed is in the folder *Example Input*. Note that the cell-ligand and cell-receptor do not have the expected filenames. Thus, you might want to change them in order to be able to read the data.

**Exercise 1a.** Use RaCInG to read the `TOY` input data. Then, answer the following questions:
- Which parts of the input data are patient specific?
- Do all patient specific input data sum to one?
- Can all cells potentially interact with each other?

We now have our input data. RaCInG can use this input data for two different things: it can compute a *kernel* $\kappa$ that can be used to derive exact cell-cell interaction statistics, but it can also use a *Monte Carlo* method to construct explicit cell-cell interaction graphs which can be used to approximate cell-cell interaction statistics. Both have their advantages and drawbacks. The kernel method is fast and exact, but it requires quite __[involved formal mathematics](https://arxiv.org/abs/2309.06066)__ to prove it works. This maths is has not been developed for all conceivable network statistics. The Monte Carlo method is significantly slower, but it can be used for all network statistics without requiring heavy mathematical machinery.

## 2. The Need for Normalisation

In the previous section you have used RaCInG to derive all outputs needed to analyse cell-cell interaction networks. At the beginning of the Results section, the RaCInG article stresses the need for normalision. In the kernel method, this can be done by calculating a version of the kernel with "normal" input data, and a version of the kernel with uniform input data. By dividing the two kernels, you can get normalized kernel values. In this exercise, we explore the need for this type of normalisation. We denote the kernel between two cell-types $x$ and $y$ by $\kappa(x, y)$, and the cell-type quantification of cell-type $x$ by $q_x$.

**Exercise 2a.** Make a scatterplot of the direct communication values (i.e., $\kappa(x, y)q_x q_y$ for all $x$ and $y$) versus (e.g.) the sum of the corresponding cell-type quantifications from the input. Answer the following questions:
- What do you notice from the plot? What unforseen consequences could this have for statistics later down the line?
- What is the spearman rho correlation between the unnormalized kernel values and the sum of cell-type quantification values? Is this a problem?

**Exercise 2b.** Redo **2a** with normalized direct communication values. Answer the following questions:
- Is the thing you noticed in the scatter plot of **2a** less prominent?
- What is the spearman rho correlation now? Is it better than previously?

**Exercise 2c.** For all patients, for each fixed pair of cell-types, create a boxplot of unnormalized kernel values. Then do the same for the normalized kernel values. What do you notice?

**Exercise 2d.** Take the logarithm of the normalized kernel values, and again generate the boxplots. Are the resulting boxplots more symmetric? Can you explain why/why not?

## 3. Monte Carlo versus Kernel

The Monte Carlo and Kernel method in RaCInG are two sides of the same coin. For the Monte Carlo method you have to specify the number of cells $N$ for each network that is generated. Apart from that, you need to have a large enough number of graphs to ensure you get a good overview. In theory, the values derived from the kernel method should match the ones from the Monte Carlo method as $N \to \infty$. In this exercise, we explore this relationship. We will denote the kernel between cell-type $t$ and $s$ by $\kappa(t,s)$, and the cell-type quantification of cell-type $t$ by $q_t$.

**Exercise 3a.** Use the kernels you have created to compute the number of wedges in the cell-cell interaction networks. For cell-types $x, y$ and $z$ the number of (unnormalized) wedges is given by $\kappa(x,y)\kappa(y,z)q_xq_yq_z$. Then, answer the following questions:

- Do the Monte Carlo and kernel results come closer for larger $N$?
